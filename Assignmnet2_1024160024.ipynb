```python
#QUES 1
roll_no = "1024160029"
L = [int(digit)*10 for digit in roll_no]
#i.Print L
print("Original L:",L)
#ii.Add elements
L.append(55) #The no is appended at the last position
print("After append(55) : ",L)
L.insert(3,55) #The no is appended at the 3 index and rest all shifted to right
print("After insert(3,55) :",L)
#iii.Remove Element
L.remove(55)
print("L after remove()",L)
L.pop(2)
print("L after pop()",L)
#iv.Sort List
L.sort()
print("Sorted List : ",L)
L.sort(reverse = True)
print("Sorted List (Descending) :",L)
#v.Slicing
print("Frist Three elements : ",L[:3])
print("Last three elements :", L[-3:])
# vi. List comprehension
average = sum(L) / len(L)
greater_than_average = [x for x in L if x > average]
print("Average:", average)
print("Elements greater than average:", greater_than_average)
```

```python
#Ques 2
# Create tuple using the first 8 elements of L
scores = tuple(L[:8])
print("Scores tuple:", scores)
# i. Highest score and lowest score
highest = max(scores)
lowest = min(scores)
print("Highest score:", highest)
print("Index of highest score:", scores.index(highest))
print("Lowest score:", lowest)
print("Lowest score appears", scores.count(lowest), "time(s)")
# ii. Reverse the tuple and return as a list
# Tuples are immutable, so they cannot be reversed in place.
reversed_list = list(scores[::-1])
print("Reversed as list:", reversed_list)
# iii. Search for a score
search = int(input("Enter a score to search: "))
if search in scores:
    print("First occurrence index:", scores.index(search))
else:
    print("Score not present in the tuple.")
# iv. Attempt to modify the tuple
try:
    scores[0] = 100
except TypeError as e:
    print("Error:", e)
# Tuples are immutable, so their elements cannot be changed after creation. Lists are mutable.
# v. Tuple unpacking
first, second, *remaining = scores
print("First score:", first)
print("Second score:", second)
print("Remaining scores:", remaining)
```

```python
#Ques 3
import random
# Set seed using your roll number
random.seed(1024160029)
# i. Generate 100 random numbers between 100 and 900
numbers = [random.randint(100, 900) for _ in range(100)]
print("Random Numbers:")
print(numbers)
# ii. Count and print odd numbers
odd_numbers = [num for num in numbers if num % 2 != 0]
print("\nOdd Numbers:", odd_numbers)
print("Count of Odd Numbers:", len(odd_numbers))
# iii. Count and print even numbers
even_numbers = [num for num in numbers if num % 2 == 0]
print("\nEven Numbers:", even_numbers)
print("Count of Even Numbers:", len(even_numbers))
# iv. Count and print prime numbers
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
prime_numbers = [num for num in numbers if is_prime(num)]
print("\nPrime Numbers:", prime_numbers)
print("Count of Prime Numbers:", len(prime_numbers))
# v. Find the most frequent number
most_frequent = max(set(numbers), key=numbers.count)
frequency = numbers.count(most_frequent)
print("\nMost Frequent Number:", most_frequent)
print("Frequency:", frequency)
```

```python
#Ques 4
# First 8 digits of the roll number
roll_no = "1024160029"
digits = [int(d) for d in roll_no[:8]]
# Create sets
A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}
print("Set A:", A)
print("Set B:", B)
# vi. Union
print("\nUnion of A and B:", A.union(B))
# vii. Intersection
print("Intersection of A and B:", A.intersection(B))
# viii. Difference
print("A - B:", A.difference(B))
print("B - A:", B.difference(A))
# difference() returns elements present in one set only.
# symmetric_difference() returns elements present in either set, but not in both.
# ix. Symmetric Difference
print("Symmetric Difference:", A.symmetric_difference(B))
# x. Subset and Superset
print("\nIs A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))
# xi. Remove a value using discard()
X = int(input("\nEnter a value to remove from Set A: "))
A.discard(X)
print("Set A after discard:", A)
# discard() does not raise an error if the value is absent, unlike remove().
```

```python
#Ques 5
my_dict = {
    "name": "Naman Arora",
    "roll_no": "1024160029",
    "branch": "Computer Science",
    "age": 20,
    "city": "Mohali"
}
print("Original Dictionary:")
print(my_dict)
# i. Rename "city" to "location" using pop()
my_dict["location"] = my_dict.pop("city")
print("\nAfter renaming city to location:")
print(my_dict)
# ii. Add CGPA
my_dict["cgpa"] = 9.51
print("\nAfter adding CGPA:")
print(my_dict)
# iii. Increase age by 1
my_dict["age"] += 1
print("\nAfter increasing age:")
print(my_dict)
# iv. Delete "branch" using pop()
dict_pop = my_dict.copy()
removed_value = dict_pop.pop("branch")
print("\nAfter pop():")
print(dict_pop)
print("Removed value:", removed_value)
# Delete "branch" using del
dict_del = my_dict.copy()
del dict_del["branch"]
print("\nAfter del:")
print(dict_del)
# pop() returns the removed value, while del only deletes the key and returns nothing.
# v. Iterate using items()
print("\nDictionary Items:")
for key, value in my_dict.items():
    print(f"{key} → {value}")
# vi. Check if "email" exists
print("\nChecking for email:")
if "email" in my_dict:
    print(my_dict["email"])
else:
    print("Email key does not exist.")
# vii. Create friend's dictionary
friend_dict = {
    "name": "Rahul Dadwal",
    "roll_no": "1024160014",
    "branch": "Computer Science",
    "age": 20,
    "city": "Phatankot"
}
merged_dict = {**my_dict, **friend_dict}
print("\nFriend Dictionary:")
print(friend_dict)
print("\nMerged Dictionary:")
print(merged_dict)
# When keys are the same, the values from the second dictionary overwrite the first.
# viii. Dictionary comprehension for string values
string_dict = {k: v for k, v in my_dict.items() if isinstance(v, str)}
print("\nDictionary containing only string values:")
print(string_dict)
```
