```python
# Assignment 5.2: WAP to add all numbers divisible by 7 and 9 from 1 to n and n is given by the user.
n = int(input("Enter a number (n): "))
total = 0

for i in range(1, n + 1):
    if i % 7 == 0 and i % 9 == 0: 
        total += i
        
print("Sum of numbers divisible by 7 and 9 from 1 to", n, "is:", total)
```

```python
# Assignment 5.3: WAP to add all prime numbers from 1 to n and n is given by the user.
n = int(input("Enter a number (n): "))
prime_sum = 0

for num in range(2, n + 1):
    is_prime = True
    for i in range(2, (num // 2) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        prime_sum += num

print("Sum of all prime numbers from 1 to", n, "is:", prime_sum)
```

```python
# Assignment 6.1: WAP using function that add all odd numbers from 1 to n, n is given by the user.
def sum_odd_numbers(n):
    total = 0
    for i in range(1, n + 1, 2):
        total += i
    return total

n = int(input("Enter a number (n): "))
print("Sum of odd numbers from 1 to", n, "-->", sum_odd_numbers(n))
```

```python
# Assignment 6.2: WAP using function that add all prime numbers from 1 to n, n given by the user.
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, (num // 2) + 1):
        if num % i == 0:
            return False
    return True

def sum_prime_numbers(n):
    total = 0
    for i in range(1, n + 1):
        if is_prime(i):
            total += i
    return total

n = int(input("Enter a number (n): "))
print("Sum of prime numbers from 1 to", n, "-->", sum_prime_numbers(n))
```
